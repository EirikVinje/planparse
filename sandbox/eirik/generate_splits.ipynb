```python
import json
import os
import numpy as np
from sklearn.model_selection import train_test_split

path = "../../formated_data/huge/extended_generated_with_edge_cases.jsonl"

with open(path, "r") as f:
    data = [json.loads(line) for line in f]

train_size = 0.7  # 70% for training
test_size = 0.2   # 20% for testing
eval_size = 0.1   # 10% for evaluation

# shuffle the data
np.random.shuffle(data)

train_data, temp_data = train_test_split(data, test_size=(1 - train_size), random_state=42)

test_proportion = test_size / (test_size + eval_size)

test_data, eval_data = train_test_split(temp_data, test_size=(1 - test_proportion), random_state=42)

save_path = "../../formated_data/huge/splits/"

with open(save_path + "train.jsonl", "w") as f:
    for item in train_data:
        f.write(json.dumps(item) + "\n")

with open(save_path + "test.jsonl", "w") as f:
    for item in test_data:
        f.write(json.dumps(item) + "\n")

with open(save_path + "eval.jsonl", "w") as f:
    for item in eval_data:
        f.write(json.dumps(item) + "\n")
```
