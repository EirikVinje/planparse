```python
import spacy

nlp = spacy.blank("nb")

nlp.add_pipe("sentencizer")

text = "Hei! Hvordan går det? Dette er en test.Jeg håper det fungerer."

# Process the text
doc = nlp(text)

# Extract phrases/sentences
phrases = [sent.text for sent in doc.sents]
print(phrases)
```
